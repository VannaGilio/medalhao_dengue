```python
# converter datas
colunas_data = [
    "data_primeiros_sintomas",
    "data_investigacao",
    "data_chikungunya_s1",
    "data_chikungunya_s2",
    "data_prnt",
    "data_sorologia",
    "data_ns1",
    "data_isolamento_viral",
    "data_pcr",
    "data_internacao",
    "data_obito",
    "data_encerramento",
    "data_sinais_alarme",
    "data_gravidade",
    "data_digitacao"
]

for coluna in colunas_data:
    df[coluna] = pd.to_datetime(
        df[coluna],
        errors="coerce"
    )

# códigos para texto
colunas_codigo = [
    "id_agravo",
    "id_municipio_notificacao",
    "id_regional",
    "id_unidade",
    "id_municipio_residencia",
    "id_regional_residencia",
    "id_pais",
    "id_ocupacao",
    "uf_notificacao",
    "uf",
    "uf_infeccao",
    "pais_infeccao",
    "municipio_infeccao"
]

for coluna in colunas_codigo:
    df[coluna] = df[coluna].astype("string")

# variáveis categóricas
colunas_categoricas = [
    "tipo_notificacao",
    "sexo",
    "gestante",
    "raca",
    "escolaridade",
    "uf",
    "classificacao_final",
    "hospitalizacao",
    "evolucao",
    "criterio_confirmacao"
]

for coluna in colunas_categoricas:
    df[coluna] = df[coluna].astype("category")

df["idade"] = pd.to_numeric(
    df["idade"],
    errors="coerce"
).astype("Int64")

df["ano"] = pd.to_numeric(
    df["ano"],
    errors="coerce"
).astype("Int64")

df["ano_nascimento"] = pd.to_numeric(
    df["ano_nascimento"],
    errors="coerce"
).astype("Int64")

print(df.dtypes)
```

```python
import matplotlib.pyplot as plt
```

1. Distribuição dos casos por faixa etária

```python
# criando faixas etárias
df_dengue["faixa_etaria"] = pd.cut(
    df_dengue["idade"],
    bins=[0, 9, 19, 29, 39, 49, 59, 69, 79, 89, 200],
    labels=[
        "0-9",
        "10-19",
        "20-29",
        "30-39",
        "40-49",
        "50-59",
        "60-69",
        "70-79",
        "80-89",
        "90+"
    ]
)

faixas = df_dengue["faixa_etaria"].value_counts().sort_index()

plt.figure(figsize=(10, 5))
faixas.plot(kind="bar")

plt.title("Distribuição dos casos por faixa etária")
plt.xlabel("Faixa etária")
plt.ylabel("Número de casos")
plt.xticks(rotation=0)
plt.show()
```

2. Estados com maior número de casos

```python
casos_uf = (
    df_dengue["uf"]
    .value_counts()
    .head(10)
    .sort_values()
)

plt.figure(figsize=(10, 6))
casos_uf.plot(kind="barh")

plt.title("Top 10 estados com maior número de casos")
plt.xlabel("Número de casos")
plt.ylabel("UF")
plt.show()
```

Municípios com maior número de casos

```python
casos_municipio = (
    df_dengue["id_municipio_residencia"]
    .value_counts()
    .head(10)
    .sort_values()
)

plt.figure(figsize=(10, 6))
casos_municipio.plot(kind="barh")

plt.title("Top 10 municípios com maior número de casos")
plt.xlabel("Número de casos")
plt.ylabel("Município")
plt.show()
```

3. Distribuição dos casos ao longo dos anos

```python
casos_ano = df_dengue["ano"].value_counts().sort_index()

plt.figure(figsize=(10, 5))
casos_ano.plot(kind="line", marker="o")

plt.title("Distribuição dos casos ao longo dos anos")
plt.xlabel("Ano")
plt.ylabel("Número de casos")
plt.xticks(casos_ano.index)
plt.grid()
plt.show()
```

4. Distribuição dos casos por mês

```python
df_dengue["mes"] = df_dengue["data_primeiros_sintomas"].dt.month

casos_mes = df_dengue["mes"].value_counts().sort_index()

nomes_meses = {
    1: "Jan",
    2: "Fev",
    3: "Mar",
    4: "Abr",
    5: "Mai",
    6: "Jun",
    7: "Jul",
    8: "Ago",
    9: "Set",
    10: "Out",
    11: "Nov",
    12: "Dez"
}

casos_mes.index = casos_mes.index.map(nomes_meses)

plt.figure(figsize=(15, 5))
casos_mes.plot(kind="line", marker="o")

plt.title("Distribuição dos casos por mês")
plt.xlabel("Mês")
plt.ylabel("Número de casos")
plt.grid()
plt.show()
```

5. Sintomas mais frequentes

```python
sintomas = [
    "febre",
    "mialgia",
    "cefaleia",
    "vomito",
    "dor_costas",
    "nausea",
    "dor_retro_orbital"
]

quantidade_sintomas = {}

for sintoma in sintomas:
    quantidade_sintomas[sintoma] = (df_dengue[sintoma] == 1).sum()

sintomas_freq = pd.Series(quantidade_sintomas).sort_values()

plt.figure(figsize=(10, 6))
sintomas_freq.plot(kind="barh")

plt.title("Sintomas mais frequentes")
plt.xlabel("Número de casos")
plt.ylabel("Sintoma")
plt.show()
```

6. Perfil dos casos graves

```python
print(df_dengue["classificacao_final"].value_counts(dropna=False))

df_graves = df_dengue[
    df_dengue["classificacao_final"] == 11
]
```

Perfil por faixa etária

```python
graves_faixa = (
    df_graves["idade"]
    .value_counts()
    .sort_index()
)

plt.figure(figsize=(10, 5))
graves_faixa.plot(kind="bar")

plt.title("Casos graves por faixa etária")
plt.xlabel("Faixa etária")
plt.ylabel("Número de casos")
plt.xticks(rotation=0)
plt.show()
```

Perfil por sexo

```python
graves_sexo = df_graves["sexo"].value_counts()

plt.figure(figsize=(7, 5))
graves_sexo.plot(kind="bar")

plt.title("Casos graves por sexo")
plt.xlabel("Sexo")
plt.ylabel("Número de casos")
plt.xticks(rotation=0)
plt.show()
```

7. Sintomas mais frequentes entre os casos graves

```python
quantidade_sintomas_graves = {}

for sintoma in sintomas:
    quantidade_sintomas_graves[sintoma] = (
        df_graves[sintoma] == 1
    ).sum()

sintomas_graves = pd.Series(
    quantidade_sintomas_graves
).sort_values()

plt.figure(figsize=(10, 6))
sintomas_graves.plot(kind="barh")

plt.title("Sintomas mais frequentes entre os casos graves")
plt.xlabel("Número de casos graves")
plt.ylabel("Sintoma")
plt.show()
```

9. Óbitos por ano

```python
print(df_dengue["evolucao"].value_counts(dropna=False))

df_obitos = df_dengue[
    df_dengue["evolucao"] == 2
]
```

Óbitos por ano

```python
obitos_ano = df_obitos["ano"].value_counts().sort_index()

plt.figure(figsize=(10, 5))
obitos_ano.plot(kind="bar")

plt.title("Óbitos por ano")
plt.xlabel("Ano")
plt.ylabel("Número de óbitos")
plt.xticks(rotation=0)
plt.show()
```

15. Sintomas segundo sexo

```python
tabela_sexo_sintomas = pd.DataFrame()

for sintoma in sintomas:
    tabela_sexo_sintomas[sintoma] = (
        df_dengue.groupby("sexo", observed=True)[sintoma]
        .apply(lambda x: (x == 1).mean() * 100)
    )

plt.figure(figsize=(12, 5))

plt.imshow(
    tabela_sexo_sintomas,
    aspect="auto"
)

plt.colorbar(label="Percentual (%)")

plt.xticks(
    range(len(tabela_sexo_sintomas.columns)),
    tabela_sexo_sintomas.columns,
    rotation=45
)

plt.yticks(
    range(len(tabela_sexo_sintomas.index)),
    tabela_sexo_sintomas.index
)

plt.title("Ocorrência dos sintomas segundo sexo")
plt.xlabel("Sintoma")
plt.ylabel("Sexo")

plt.show()
```

12. Municípios com maior proporção de casos graves

```python
total_municipio = (
    df_dengue["id_municipio_residencia"]
    .value_counts()
)

graves_por_municipio = (
    df_graves["id_municipio_residencia"]
    .value_counts()
)

proporcao_graves = (
    graves_por_municipio / total_municipio * 100
).dropna()

proporcao_graves = (
    proporcao_graves
    .sort_values(ascending=False)
    .head(10)
    .sort_values()
)

plt.figure(figsize=(10, 6))
proporcao_graves.plot(kind="barh")

plt.title("Top 10 municípios por proporção de casos graves")
plt.xlabel("Casos graves (%)")
plt.ylabel("Município")
plt.show()
```
